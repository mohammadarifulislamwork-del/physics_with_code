# A cart starts at x = 0 meters with a velocity of 0.45 m/s at a time of t = 0 seconds. If the cart has an acceleration of 0.2 m/s2, where will the cart be after 1.5 seconds?

```python
x = 0 #m
v = 0.45 #m/s
t = 0 #s
a = 0.2 #m/s^2
dt = 0.1 #s

while t<1.5:
  v = v + a*dt
  x = x + v*dt 
  t = t + dt

print("x = ",x,"m")
```
